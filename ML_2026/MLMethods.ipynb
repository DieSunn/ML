penalty:

l1 - Применяет L1-регуляризацию (лассо). Добавляет к функции потерь сумму абсолютных значений коэффициентов, умноженную на C (обратную силу регуляризации). Заставляет некоторые коэффициенты стать равными нулю — фактически выполняет отбор признаков.

l2 - Применяет L2-регуляризацию (гребневую). Добавляет сумму квадратов коэффициентов. Коэффициенты становятся маленькими, но редко обнуляются. Это самый распространённый тип регуляризации.

elasticnet - Комбинация L1 и L2. Имеет дополнительный гиперпараметр l1_ratio, который управляет балансом между L1 и L2.

none - Без регуляризации. Модель обучается минимизировать только функцию потерь.

l1_ration - имеет смысл использовать, когда l1 и l2 используются.

# Функция для обучения

```python
from sklearn.base import BaseEstimator
import numpy as np

def fit_and_test(model: BaseEstimator, X_train:np.ndarray, X_test:np.ndarray, y_train:np.ndarray, y_test: np.ndarray):
    """Обучает model и печатает таблицы с метриками качества
    Возвращает accuracy_train, accuracy_test
    """
    model.fit(X_train, y_train)

    y_train_pred = model.predict( X_train )
    y_test_pred = model.predict( X_test )

    print("Train:")
    print(classification_report(y_train, y_train_pred))

    print("\n\nTest:")
    print(classification_report(y_test, y_test_pred))

    return accuracy_score(y_train, y_train_pred), accuracy_score(y_test, y_test_pred)
```

# Функция для сохранения в mlflow

```python
import mlflow
import mlflow.sklearn
from sklearn.model_selection import GridSearchCV
from sklearn.metrics import accuracy_score

def train_and_log_random_search(
    estimator,
    param_grid,
    X_train,
    y_train,
    X_test,
    y_test,
    run_name="ML_Run",
    model_type="Model",
    scaling_method="None",
    cv=5
):
    """
    Функция для проведения RandomizedSearchCV и логирования результатов в MLflow.
    """

    # Настройка RandomizedSearchCV
    random_search = GridSearchCV(
        estimator=estimator,
        param_grid=param_grid,
        cv=cv,
        scoring='accuracy',
        n_jobs=-1,
        verbose=1
    )

    # Обучение
    random_search.fit(X_train, y_train)

    # Предсказание и метрики
    y_pred_train = random_search.predict(X_train)
    y_pred_test = random_search.predict(X_test)

    acc_train = accuracy_score(y_train, y_pred_train)
    acc_test = accuracy_score(y_test, y_pred_test)

    # Логирование в MLflow
    with mlflow.start_run(run_name=run_name):

        # Информация о данных
        mlflow.log_params({
            "train_size": len(y_train),
            "test_size": len(y_test),
            "class_balance": float(y_train.mean()),
            "features_n": X_train.shape[1],
            "scaling": scaling_method,
            "model_type": model_type
        })

        # Метрики
        mlflow.log_metrics({
            "acc_train": acc_train,
            "acc_test": acc_test,
            "best_cv_score": random_search.best_score_
        })

        # Параметры поиска (превращаем в строку, если это словарь)
        mlflow.log_param("param_distributions", str(param_dist))
        mlflow.log_param("best_params", str(random_search.best_params_))

        # Сохранение лучшей модели
        mlflow.sklearn.log_model(random_search.best_estimator_, artifact_path="model")

    return random_search
```

1) Устройство catboosting

Сначала модель делает очень простое предсказание (например, всегда «среднее значение»).

Потом она строит первое дерево, которое учится исправлять ошибки этого простого предсказания.

Затем добавляется ещё одно дерево, которое учится исправлять ошибки уже двух предыдущих моделей, и так по кругу.
В итоге финальный ответ — сумма всех этих деревьев, и каждый новый шаг только «подправляет» то, что предыдущие промахнулись

$\hspace{1cm}$

2)
CatBoost сам умеет работать с категориями (str, названия городов, категорий и т.п.), не требуя отдельного One‑Hot или LabelEncoder.

Для каждой категории он считает статистику (например, какая доля целевой переменной в этой группе) и делает из этого численный признак‑счётчик.

Это удобно и часто даёт лучшее качество, чем ручное кодирование.

$\hspace{1cm}$

3) Борьба с переобучением
CatBoost строит не слишком глубокие деревья и использует регуляризацию, чтобы не «подстраиваться» слишком прямо под обучающую выборку.

Он ещё применяет случайные перестановки данных и другие техники, чтобы модели было сложнее запомнить шум.

## Плюсы

- Высокая точность и устойчивость к переобучению: CatBoost как базовый estimator наследует ordered boosting и регуляризацию, что даёт лучшую обобщающую способность по сравнению с чистым sklearn бустингом (accuracy ~0.807 для sklearn vs выше для CatBoost)

- Обработка категориальных данных и пропусков: Нет нужды в one-hot или imputation — CatBoost это делает автоматически, упрощая пайплайн в sklearn ансамблях.

- Интеграция с экосистемой sklearn: Легко комбинировать с GridSearchCV, Pipeline и другими инструментами для тюнинга и валидации

## Минусы:

- Сложность настройки и нестабильность: Много гиперпараметров CatBoost усложняет конфигурацию, особенно с несбалансированными данными; документация слаба

- Потенциальная потеря скорости: CatBoost медленнее чистого HistGradientBoosting на больших данных, а в ансамбле добавляется overhead от sklearn wrapper (n_estimators=1 для тестов).

- Несовместимость и overhead: GradientBoosting sklearn не предназначен для внешних бустеров вроде CatBoost; HistGradientBoosting лучше, но требует оборачивания, что может снижать производительность и вызывать ошибки

Основные функции валидации:

- Подбор гиперпараметров: Валидация помогает настраивать параметры, которые не корректируются в процессе обучения (например, глубина дерева, скорость обучения или коэффициенты регуляризации).

- Контроль переобучения: Если ошибка на обучающей выборке продолжает падать, а на валидационной начинает расти, это явный признак того, что модель начинает «запоминать» шум в данных, а не находить общие закономерности.

- Early Stopping (ранняя остановка): Валидационная выборка позволяет автоматически прекратить процесс обучения в момент достижения наилучшего качества, чтобы предотвратить деградацию модели.
